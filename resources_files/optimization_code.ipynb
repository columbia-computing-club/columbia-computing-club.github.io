# **Linear Programming**

a) 2x + 3y + 5z >= 100

b) 3x + 4y + 10z >= 200

c) 5x + 5y + 11z <= 250

Goal: Maximize 10x + 10y + z

```python
from pulp import LpMaximize, LpProblem, LpVariable, lpSum, LpStatus, value
from pulp import *
# Creating the LP problem
model = LpProblem(name="maximize-goal", sense=LpMaximize)
# Define variables
x = LpVariable(name="x", lowBound=0)
y = LpVariable(name="y", lowBound=0)
z = LpVariable(name="z", lowBound=0)

# Constraints and objective
model += (2*x + 3*y + 5*z >= 100, "Constraint_a")
model += (3*x + 4*y + 10*z >= 200, "Constraint_b")
model += (5*x + 5*y + 11*z <= 250, "Constraint_c")
model += 10*x + 10*y + z, "Objective"

# Solve and Output
status = model.solve()
print(f"Status: {LpStatus[status]}")
print(f"x = {x.value()}, y = {y.value()}, z = {z.value()}")
print(f"Objective value = {value(model.objective)}")

# Let us check to see if more than 1 solution exist- if alternate = previous then there is no other os.
status = model.solve()
original_obj = value(model.objective)
original_x = x.value()
original_y = y.value()
original_z = z.value()
model += (10*x + 10*y + z == original_obj, "Fix_Objective")
# model += (abs(10*x + 10*y + z - original_obj) <= 1e-6, "Fix_Objective") to add tolerance
model.setObjective(-y) # Extra objective to see other values
print("Alternate solution:")
print(f"x = {x.value()}, y = {y.value()}, z = {z.value()}")
```

# **Checking Feasibility (as a LP)**

```python
# Minimize dummy objective
prob = pulp.LpProblem("FeasibilityCheck", LpMinimize)

# Decision variables
x = pulp.LpVariable("x", lowBound=0)
y = pulp.LpVariable("y", lowBound=0)

# Dummy objective
prob += 0

# Constraints
prob += 2*x + 3*y <= 100
prob += 3*x + 4*y >= 200

# Solve
prob.solve()

# Check status
if LpStatus[prob.status] == "Optimal":
    print("Feasible")
else:
    print("Infeasible")
```

# **Integer Programming**

John's Pizza: John prepares his pizzas by getting them ready with sauce and cheese beforehand. He does this every hour and based on past trends and needs has figured that:

Cheese Pizza needs: 2 Jars of Sauce and 3 Cups of Cheese

Margherita Pizza needs: 3 Jars of Sauce 2 Cups of Cheese

1 Jar of Sauce costs USD1 and 1 cup of cheese costs USD2. A maximum of USD100 worth of product will be used for prep for any given hour.

Number of Cheese pizzas made must be greater than Mgt as former has higher demand. However, for every 3 Cheese pizzas sold 1 Mgt shall be sold. Doing this ensures no wastage and repeat customers.

Maximize Profit (given that selling a Cheese pizza has USD2 profit and Mgt has USD1.5.)

```python
# Model
model = LpProblem("Pizza_Production", LpMaximize)

# Variables
x = LpVariable("Cheese_Pizza", lowBound=0, cat="Integer")
y = LpVariable("Mgt_Pizza", lowBound=0, cat="Integer")
S = LpVariable("Sauce_Jars", lowBound=0, cat="Integer")
C = LpVariable("Cheese_Cups", lowBound=0, cat="Integer")

# Profit
pc = 2.0
pm = 1.5

# Objective
model += pc*x + pm*y

# Resource constraints
model += 2*x + 3*y <= S
model += 3*x + 2*y <= C

# "Budget" (Not really a budget more so that we don't want to use more than this much product from inventory at any given point.)
model += S + 2*C <= 100 # Can be written just in terms of x and y. Meaning you can do by hand what S + 2*C is and just make 100 as upper limit.

# Demand constraints
model += x >= y
model += x <= 3*y

# Solve
model.solve()

# Output
print("Status:", LpStatus[model.status])
print("Cheese pizzas:", x.value())
print("Mgt pizzas:", y.value())
print("Sauce jars:", S.value())
print("Cheese cups:", C.value())
print("Profit:", value(model.objective))
```

# **Multiobjective Function**

A blood drive center needs at least 300 "units" of blood and has 300 people willing to donate. They will collect blood over 60 days during the quarter. A person can donate at most 10 "units" of blood in a single visit (per day). After first donation a person can show up 2 more times during the drive.

The first time someone donates, a cost of USD 50 is incurred (data entry, screening tests, etc.). Every subsequent donation by the same person costs USD 10.

Each person has a 30% chance of being eliminated from donating after their initial screening test. However, we don't know beforehand who will be screened out meaning fixed cost per person is incurred even if they cannot participate.

There is no limit on how many people can show up on a given day. However, the center cannot collect more than 5000 "units" of blood in total due to storage capacity.

Suppose the center wishes to maximize the total blood collected while minimizing the total cost.

(Just for example- not completely realistic)

```python
n = 300 # Number of donors
eligible_rate = 0.7
eligibleMAX = int(n * eligible_rate)

days = 60
units_per_visit = 10
storage_cap = 5000
min_req = 300

first_visit_cost = 50
repeat_visit_cost = 10

weight_sets = [(10,1), (2,1), (0.5,1)] # Here we can enter 3 (or any number- just add more brackets) of weight combinations
# Important: We didn't normalize or do anything like that so (1,0.5) doesn't mean twice as important- code only shows trade-offs

for alpha, beta in weight_sets:

    # Define the model
    model = LpProblem("Blood_Drive_Optimization", LpMaximize)

    # Decision variables
    v = LpVariable.dicts("visit",
                            ((i, d) for i in range(n) for d in range(days)),
                            cat='Binary')
    fv = LpVariable.dicts("first_visit",
                            (i for i in range(n)),
                            cat='Binary')

    # Expressions
    total_blood = lpSum(units_per_visit * v[i, d]
                           for i in range(n)
                           for d in range(days))
    total_visits = lpSum(v[i, d]
                            for i in range(n)
                            for d in range(days))
    t_first_visits = lpSum(fv[i] for i in range(n))
    t_repeat_visits = total_visits - t_first_visits
    total_cost = first_visit_cost * n + repeat_visit_cost * total_visits

    # Objective
    model += alpha * total_blood - beta * total_cost

    # Constraints
    model += total_blood >= min_req
    model += total_blood <= storage_cap

    # Deterministic: We assume exactly 70% of donors are eligible (210 donors total)
    model += lpSum(v[i, d] for i in range(n) for d in range(days)) <= 3 * eligibleMAX # In total nobody can donate more than 3 times

    for i in range(n):
        model += lpSum(v[i, d] for d in range(days)) >= fv[i]
        model += lpSum(v[i, d] for d in range(days)) <= 3 * fv[i]

    # Solve
    model.solve()

    # Output
    print("Weights (Blood, Cost):", (alpha, beta))
    print("Status:", LpStatus[model.status])
    print("Total Blood Collected:", value(total_blood))
    print("Total Cost:", value(total_cost))
    print("Objective Value:", value(model.objective))
    print()
```

# **Kidney Exchange Example**
[x,y] means that donor from pair x is compatible with patient from pair y. Hence patient y can receive a kidney from donor x.
Suppose we have 15 pairs. Let us first only allow a 2 way exchange. The goal will be the maximum number of possible exchanges.

```python
data = [
    [1,10], # So donor from pair 1 can give kidney to patient 10
    [1,11],
    [1,13],
    [2,5],
    [2,6],
    [2,8],
    [3,1],
    [3,4],
    [3,5],
    [3,6],
    [3,10],
    [3,11],
    [3,13],
    [3,15],
    [4,2],
    [4,6],
    [4,9],
    [4,14],
    [5,1],
    [5,2],
    [5,4],
    [5,6],
    [6,8],
    [6,10],
    [6,12],
    [6,13],
    [7,10],
    [7,11],
    [7,12],
    [8,1],
    [9,3],
    [9,4],
    [9,5],
    [10,2], # But there is no [10,1] pair meaning 1 and 10 cannot do an exchange
    [10,4],
    [10,6],
    [10,7],
    [10,8],
    [10,12],
    [11,3],
    [11,5],
    [11,8],
    [12,9],
    [12,10],
    [13,6],
    [14,4],
    [15,3],
    [15,5],
    [15,7],
    [15,13]
]
# Create problem
kidney_problem = LpProblem("Max_Kidney_Exchanges", LpMaximize)

# Convert data into tuples (just to make them hashable)
data_tuples = set()
for pair in data:
    data_tuples.add((pair[0], pair[1]))

# Only keep valid 2-way exchanges (i.e., both directions exist)
valid_pairs = set()
for (i, j) in data_tuples:
    if (j, i) in data_tuples:
        # To avoid duplicates like (a,b) and (b,a),
        # only keep one ordering
        if i < j:
            valid_pairs.add((i, j))

# Decision variables: 1 if this 2-way exchange happens
x = {}
for (i, j) in valid_pairs:
    x[(i, j)] = LpVariable(f"x_{i}_{j}", cat=LpBinary)

# Objective: maximize number of exchanges
kidney_problem += lpSum(x[(i, j)] for (i, j) in valid_pairs)

# Each patient/donor pair can appear in at most one exchange
all_nodes = set()
for (i, j) in valid_pairs:
    all_nodes.add(i)
    all_nodes.add(j)

for node in all_nodes:
    kidney_problem += (
        lpSum(x[(i, j)] for (i, j) in valid_pairs if i == node or j == node) <= 1
    )

# Solve
kidney_problem.solve()

# Print solution
print("Optimal Kidney Exchanges:")
for (i, j) in valid_pairs:
    if value(x[(i, j)]) == 1:
        print(f"Exchange between Patient/Donor {i} and  Donor/Patient {j} respectively")
```

Let's also allow 3 way exchanges. Our goal is for maximum number of people to get a kidney. A 3 way exchange means that 3 people get a kidney. But only allowing 3 way exchanges doesn't mean that the maximum number of people will get a kidney.

```python
kidney_problem = LpProblem("Max_Kidney_Exchanges_2_3_Way", LpMaximize)

# All valid 2-way exchanges
two_way_matches = set()
for edge in data_tuples:
    reverse_edge = (edge[1], edge[0])
    # Only add (a,b) if (b,a) also exists and a < b to ensure unique representation of 2-way exchange variables
    if reverse_edge in data_tuples and edge[0] < edge[1]:
        two_way_matches.add(edge)

# All valid 3-way exchanges
three_way_cycles = set()

for a, b in data_tuples:
    for c, d in data_tuples:
        # Looking for a -> b -> d type of sequence
        if b == c and (d, a) in data_tuples: # b and c have to be same person
            # We did this because we are looking for (n1, n2), (n2, n3), (n3, n1)
            # Make sure all three are different people
            if len({a, b, d}) == 3:
                cycle = (a, b, d)
                # Rotate so smallest label comes first (avoid duplicates like (2,1,3))
                # Duplicate entry like (2,1,3) can be written as (1,3,2) so we don't need (2,1,3)
                smallest = min(cycle)
                while cycle[0] != smallest:
                    cycle = (cycle[1], cycle[2], cycle[0])

                three_way_cycles.add(cycle)

# Decision variables
# For symmetric pairs, we only added (a,b) where a<b. The variable then represents the exchange between a and b.
two_way_vars = {match: LpVariable(f"x_{match[0]}_{match[1]}", 0, 1, LpBinary) for match in two_way_matches}

# For 3-way cycles, we make ordered tuple (e.g., (1,2,3))- "triplet" representing the flow 1->2->3->1
three_way_vars = {cycle: LpVariable(f"y_{cycle[0]}_{cycle[1]}_{cycle[2]}", 0, 1, LpBinary) for cycle in three_way_cycles}

# Objective: Maximize total number of kidneys donated
# Each 2-way exchange involves 2 kidneys.
# Each 3-way exchange involves 3 kidneys.
kidney_problem += lpSum(2 * two_way_vars[match] for match in two_way_vars) + \
                  lpSum(3 * three_way_vars[cycle] for cycle in three_way_vars), "Total_Kidneys_Exchanged"

# Constraint: One pair cannot be in more than 1 exchange.
participants = set(node for edge in data_tuples for node in edge)

for person in participants:
    # Sum of 2-way exchanges where 'p' is involved
    # Since symmetric_pairs are (min,max), 'p in pair' correctly checks if p is either element.
    two_way_used = lpSum(two_way_vars[match] for match in two_way_vars if person in match)

    # Sum of 3-way exchanges where 'p' is involved
    # 'p in triplet' correctly checks if p is one of the nodes in the cycle.
    three_way_used = lpSum(three_way_vars[cycle] for cycle in three_way_vars if person in cycle)

    kidney_problem += two_way_used + three_way_used <= 1 # I don't know what to call this constraint. Basically that person is either in 2 or 3 way exchange but not both

kidney_problem.solve()

# Output
chosen_2_way = [(match[0], match[1]) for match in two_way_vars if value(two_way_vars[match]) == 1]
chosen_3_way = [(cycle[0], cycle[1], cycle[2]) for cycle in three_way_vars if value(three_way_vars[cycle]) == 1]

# Total unique patients involved (more useful later)
matched_people = set()
for match in chosen_2_way:
    matched_people.add(match[0])
    matched_people.add(match[1])
for cycle in chosen_3_way:
    matched_people.add(cycle[0])
    matched_people.add(cycle[1])
    matched_people.add(cycle[2])
count = len(matched_people)

print("\nOptimal Kidney Exchanges:")
for match in sorted(chosen_2_way):
    print(f"2-Way: {match[0]} <-> {match[1]}")
for cycle in sorted(chosen_3_way):
    print(f"3-Way: {cycle[0]} -> {cycle[1]} -> {cycle[2]} -> {cycle[0]}")

for match in sorted(chosen_2_way):
    print(f"Exchange between Patient/Donor {match[0]} and Donor/Patient {match[1]} (respectively)")
for cycle in sorted(chosen_3_way):
    print(f"Exchange between Donor {cycle[0]} and Patient {cycle[1]} AND Donor {cycle[1]} and Patient {cycle[2]} AND Donor {cycle[2]} and Patient {cycle[0]}")

print(f"In these exchanges we have: {count} unique pairs/patients")
```

```python
kidney_problem = LpProblem("Max_Kidney_Exchanges_2_3_Way", LpMaximize)
priority = 14 # Some patient who has to get kidney in this whole cycle or whatever
# All valid 2-way exchanges
two_way_matches = set()
for edge in data_tuples:
    reverse_edge = (edge[1], edge[0])
    # Only add (A,B) if (B,A) also exists and A < B to ensure unique representation of 2-way exchange variables
    if reverse_edge in data_tuples and edge[0] < edge[1]:
        two_way_matches.add(edge)

# All valid 3-way exchanges
three_way_cycles = set()

for a, b in data_tuples:
    for c, d in data_tuples:
        # Looking for a -> b -> d type of sequence
        if b == c and (d, a) in data_tuples: # b and c have to be same person
            # We did this because we are looking for (n1, n2), (n2, n3), (n3, n1)
            # Make sure all three are different people
            if len({a, b, d}) == 3:
                cycle = (a, b, d)
                # Rotate so smallest label comes first (avoid duplicates like (2,1,3))
                # Duplicate entry like (2,1,3) can be written as (1,3,2) so we don't need (2,1,3)
                smallest = min(cycle)
                while cycle[0] != smallest:
                    cycle = (cycle[1], cycle[2], cycle[0])

                three_way_cycles.add(cycle)

# Decision variables
# For symmetric pairs, we only added (A,B) where A<B. The variable then represents the exchange between A and B.
two_way_vars = {match: LpVariable(f"x_{match[0]}_{match[1]}", 0, 1, LpBinary) for match in two_way_matches}

# For 3-way cycles, we make ordered tuple (e.g., (1,2,3))- "triplet" representing the flow 1->2->3->1
three_way_vars = {cycle: LpVariable(f"y_{cycle[0]}_{cycle[1]}_{cycle[2]}", 0, 1, LpBinary) for cycle in three_way_cycles}

# Objective: Maximize total number of kidneys donated
# Each 2-way exchange involves 2 kidneys.
# Each 3-way exchange involves 3 kidneys.
kidney_problem += lpSum(2 * two_way_vars[match] for match in two_way_vars) + \
                  lpSum(3 * three_way_vars[cycle] for cycle in three_way_vars), "Total_Kidneys_Exchanged"

# Constraint: One pair cannot be in more than 1 exchange.
participants = set(node for edge in data_tuples for node in edge)

for person in participants:
    # Sum of 2-way exchanges where 'p' is involved
    # Since symmetric_pairs are (min,max), 'p in pair' correctly checks if p is either element.
    two_way_used = lpSum(two_way_vars[match] for match in two_way_vars if person in match)

    # Sum of 3-way exchanges where 'p' is involved
    # 'p in triplet' correctly checks if p is one of the nodes in the cycle.
    three_way_used = lpSum(three_way_vars[cycle] for cycle in three_way_vars if person in cycle)

    kidney_problem += (two_way_used + three_way_used <= 1, f"Participant_{person}_Max_One_Exchange")

# Force priority participant to be matched
two_way_priority = lpSum(two_way_vars[pair]
                         for pair in two_way_vars
                         if priority in pair)

three_way_priority = lpSum(three_way_vars[triplet]
                           for triplet in three_way_vars
                           if priority in triplet)

kidney_problem += two_way_priority + three_way_priority == 1, \
                  f"Priority_{priority}_Must_Be_Matched"


kidney_problem.solve()

# Output
chosen_2_way = [(match[0], match[1]) for match in two_way_vars if value(two_way_vars[match]) == 1]
chosen_3_way = [(cycle[0], cycle[1], cycle[2]) for cycle in three_way_vars if value(three_way_vars[cycle]) == 1]

# Total unique patients involved (more useful now- if it has decreased we know the prioritizing was a 'bad' idea)
matched_people = set()
for match in chosen_2_way:
    matched_people.add(match[0])
    matched_people.add(match[1])
for cycle in chosen_3_way:
    matched_people.add(cycle[0])
    matched_people.add(cycle[1])
    matched_people.add(cycle[2])
count = len(matched_people)

print("\nOptimal Kidney Exchanges:")
for match in sorted(chosen_2_way):
    print(f"2-Way: {match[0]} <-> {match[1]}")
for cycle in sorted(chosen_3_way):
    print(f"3-Way: {cycle[0]} -> {cycle[1]} -> {cycle[2]} -> {cycle[0]}")

for match in sorted(chosen_2_way):
    print(f"Exchange between Patient/Donor {match[0]} and Donor/Patient {match[1]} (respectively)")
for cycle in sorted(chosen_3_way):
    print(f"Exchange between Donor {cycle[0]} and Patient {cycle[1]} AND Donor {cycle[1]} and Patient {cycle[2]} AND Donor {cycle[2]} and Patient {cycle[0]}")

print(f"In these exchanges we have: {count} unique pairs/patients")

print(f"Status: {LpStatus[kidney_problem.status]}")
```
